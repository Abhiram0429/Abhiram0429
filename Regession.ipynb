```python
df = df.dropna() #removes rows with nulls anywhere
df= df.dropna(subset=['Column_Name'], inplace=True) # removes column
df['Age'] = df['Age'].fillna(df['Age'].mean()) #Fill with the mean numerical columns
df['City'] = df['City'].fillna(df['City'].mode()[0]) #Useful for categorical columns
```

```python
#X → INPUTS → What the model learns from
#y → OUTPUT → What the model predicts
x = df[[""]]
y = df[""]
```

```python
#displays x coloums
 x
```

```python
#displays y coloums
y
```

```python
# Splits X and y into training data (80%) and testing data (20%)
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.20)
```

```python
# Train the model using the training data
# x_train = input data
# y_train = actual answers
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train,y_train)
```

```python
pred_y = lr.predict(x_test)
```

```python
from sklearn import metrics
print("MAE:", metrics.mean_absolute_error(y_test, pred_y))
print("MSE:", metrics.mean_squared_error(y_test, pred_y))
print("RMSE:", np.sqrt(metrics.mean_squared_error(y_test, pred_y)))
print("R2:", metrics.r2_score(y_test, pred_y))
```

```python
#DecisionTreeRegressor
------------------
from sklearn.tree import DecisionTreeRegressor
dtr = DecisionTreeRegressor(random_state=42)
dtr.fit(x_train, y_train)
pred_dtr = dtr.predict(x_test)
```

```python
#r2 score
from sklearn import metrics
print("MAE:", metrics.mean_absolute_error(y_test, pred_dtr))
print("MSE:", metrics.mean_squared_error(y_test, pred_dtr))
print("RMSE:", np.sqrt(metrics.mean_squared_error(y_test, pred_dtr)))
print("R2:", metrics.r2_score(y_test, pred_dtr))
```

```python
#RandomForestRegressor
------------------
from sklearn.ensemble import RandomForestRegressor
rf = RandomForestRegressor(random_state=42)
rf.fit(x_train, y_train)
pred_rf = rf.predict(x_test)
```

```python
#r2 score
from sklearn import metrics
print("MAE:", metrics.mean_absolute_error(y_test, pred_rf))
print("MSE:", metrics.mean_squared_error(y_test, pred_rf))
print("RMSE:", np.sqrt(metrics.mean_squared_error(y_test, pred_rf)))
print("R2:", metrics.r2_score(y_test, pred_rf))
```

```python
#svm
from sklearn.svm import SVR
svm = SVR()
svm.fit(x_train, y_train)
pred_svc = svm.predict(x_test)
```

```python
#r2 score
from sklearn import metrics
print("MAE:", metrics.mean_absolute_error(y_test, pred_svc))
print("MSE:", metrics.mean_squared_error(y_test, pred_svc))
print("RMSE:", np.sqrt(metrics.mean_squared_error(y_test, pred_svc)))
print("R2:", metrics.r2_score(y_test, pred_svc))
```

```python
from sklearn.svm import SVR
from sklearn.model_selection import RandomizedSearchCV, GridSearchCV

# Create SVM Regression model
svm = SVR()

# ---------------- RANDOM SEARCH ----------------

random_params = {
    'C': [0.1, 1, 10, 100],
    'kernel': ['linear', 'rbf', 'poly'],
    'gamma': ['scale', 'auto'],
    'epsilon': [0.01, 0.1, 0.2]
}

random_search = RandomizedSearchCV(
    svm,
    param_distributions=random_params,
    n_iter=10,
    cv=5,
    scoring='r2',
    random_state=42
)

random_search.fit(x_train, y_train)

print("===== RANDOM SEARCH =====")
print("Best Parameters:", random_search.best_params_)
print("Best R2 Score:", random_search.best_score_)


# ---------------- GRID SEARCH ----------------

grid_params = {
    'C': [1, 10, 100],
    'kernel': ['rbf', 'linear'],
    'gamma': ['scale', 'auto'],
    'epsilon': [0.01, 0.1]
}

grid_search = GridSearchCV(
    svm,
    param_grid=grid_params,
    cv=5,
    scoring='r2'
)

grid_search.fit(x_train, y_train)

print("\n===== GRID SEARCH =====")
print("Best Parameters:", grid_search.best_params_)
print("Best R2 Score:", grid_search.best_score_)
```

```python
#Polynomial
from sklearn.preprocessing import PolynomialFeatures
poly = PolynomialFeatures(degree = 2) # this is only for the adding x varable to get better result
poly_fea = poly.fit_transform(x)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(poly_fea,y,test_size = 0.2)
```

```python
from sklearn.linear_model import LinearRegression
lm = LinearRegression()
lm.fit(x_train,y_train)
```

```python
pred_y=lm.predict(x_test)
```

```python
#r2 score
from sklearn import metrics
print("MAE:", metrics.mean_absolute_error(y_test, pred_y))
print("MSE:", metrics.mean_squared_error(y_test, pred_y))
print("RMSE:", np.sqrt(metrics.mean_squared_error(y_test, pred_y)))
print("R2:", metrics.r2_score(y_test, pred_y))
```

```python
from sklearn.metrics import r2_score

# Linear Regression
y_pred_lr = lr.predict(x_test)
r2_lr = r2_score(y_test, y_pred_lr)

# SVM
y_pred_svm = svm.predict(x_test)
r2_svm = r2_score(y_test, y_pred_svm)

# Random Search SVM
y_pred_random = random_search.predict(x_test)
r2_random = r2_score(y_test, y_pred_random)

# Grid Search SVM
y_pred_grid = grid_search.predict(x_test)
r2_grid = r2_score(y_test, y_pred_grid)

# Polynomial Regression
y_pred_poly = poly.predict(x_test)
r2_poly = r2_score(y_test, y_pred_poly)


# Display all R2 scores
print("===== R2 SCORE COMPARISON =====")
print("Linear Regression      :", r2_lr)
print("SVM                    :", r2_svm)
print("Random Search SVM      :", r2_random)
print("Grid Search SVM        :", r2_grid)
print("Polynomial Regression  :", r2_poly)
```
